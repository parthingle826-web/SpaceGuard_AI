```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set project paths
ROOT_DIR = Path("..").resolve()
DATA_PATH = ROOT_DIR / "backend" / "data" / "satellite_telemetry_synthetic.csv"

print(f"Project root: {ROOT_DIR}")
print(f"Dataset exists: {DATA_PATH.exists()}")
```

### 1. Ingest Dataset & Inspect Schema

```python
df = pd.read_csv(DATA_PATH)
print(f"Total Observations: {len(df):,}")
print(f"Features Count: {df.shape[1]}")
df.head()
```

### 2. Anomaly Class Distribution

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, y="anomaly_type", order=df["anomaly_type"].value_counts().index, palette="mako")
plt.title("SpaceGuard AI: Anomaly Category Distribution")
plt.xlabel("Count")
plt.ylabel("Anomaly Category")
plt.tight_layout()
plt.show()
```

### 3. Correlation Matrix of Subsystem Telemetry

```python
features = [
    "battery_voltage", "battery_current", "temperature", "pressure",
    "solar_panel_voltage", "solar_panel_current", "power_consumption",
    "cpu_subsystem_temp", "signal_strength", "radiation_level"
]
corr = df[features].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Correlation Matrix of Satellite Subsystem Features")
plt.tight_layout()
plt.show()
```
